## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
df = sns.load_dataset('tips')
df.info()
print(df.shape)
df.head()
```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
df.select_dtypes(include='float').hist(figsize=(9,6), bins=15)
plt.show()
sns.boxplot(data=df.select_dtypes(include='float'))
plt.title('Deteksi Outlier'); plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# TULIS KODE ANDA DI SINI (Latihan 3)
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')
plt.show()
sns.pairplot(df, hue='time'); plt.show()
```

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# TULIS KODE ANDA DI SINI (Latihan 4)
num = df.select_dtypes(include='number')
sns.heatmap(num.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```

1. Mengapa EDA penting dilakukan sebelum membangun model?
2. Apa perbedaan informasi yang Anda peroleh dari histogram, box plot, dan scatter plot?
3. Temuan EDA mana pada dataset Iris yang menurut Anda paling berguna untuk klasifikasi? Mengapa?

---
*Jawaban Anda:*

1. EDA penting dilakukan karena sebelum membangun model, kita perlu mengetahui kondisi data terlebih dahulu. Dalam supervised learning, terdapat input dan label, dan jika datanya masih bermasalah, seperti adanya nilai kosong (NaN) atau outlier, maka model bisa belajar dari data yang salah atau bias. Nilai kosong dapat dicek menggunakan `df.isnull().sum()`. Jika jumlahnya sedikit, nilai tersebut bisa diisi (imputasi) menggunakan nilai minimum, median, atau modus. Namun jika jumlah nilai kosongnya banyak, lebih baik fitur tersebut dihapus saja. Outlier juga perlu ditelaah terlebih dahulu, apakah itu memang kesalahan data atau merupakan data valid yang nilainya ekstrem, sebelum diputuskan untuk dihapus. Jadi, EDA berfungsi untuk memastikan data sudah layak digunakan sebelum masuk ke tahap pemodelan.

2. - **Histogram** menunjukkan sebaran atau distribusi dari satu fitur numerik, misalnya apakah data mengumpul di tengah atau miring ke salah satu sisi.
- **Box plot** menunjukkan median, kuartil (25%, 50%, dan 75%), serta digunakan untuk mendeteksi outlier.
- **Scatter plot** menunjukkan hubungan antara dua fitur numerik, apakah keduanya saling berkaitan (naik atau turun bersamaan) atau tidak ada hubungan sama sekali.

Jadi, ketiganya memiliki fokus yang berbeda: histogram untuk satu variabel, scatter plot untuk dua variabel, dan box plot untuk ringkasan statistik sekaligus deteksi outlier.

3. Temuan yang paling berguna adalah total_bill dan tip punya hubungan positif, semakin besar tagihan, semakin besar tipnya. Ini berguna karena total_bill bisa jadi fitur kuat untuk memprediksi tip.
